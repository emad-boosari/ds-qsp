## Signal Generation
A multi-Gaussian signal is generated as a representative multi-peak localized waveform for transform-domain sparsity analysis.

```python
n = 10 # Number of qubit
N = 2**n  # Number of quantum basis

x  = np.linspace(-5, 5, N)
rng = np.random.default_rng(42)  # reproducible example

K = 12  # number of Gaussians
mus    = rng.uniform(-4.5, 4.5, size=K)   # centers
sigmas = rng.uniform(0.12, 0.60, size=K)  # widths
amps   = rng.uniform(0.30, 1.00, size=K)  # amplitudes

g = np.zeros_like(x)
for a, m, s in zip(amps, mus, sigmas):
    g += a * np.exp(-(x - m)**2 / (2.0 * s**2))

# small noise
g += 0.001 * rng.standard_normal(N)

target_f =  g / np.linalg.norm(g)
```

```python
# ============================================================
# DFT compression of Single Gaussian
# ============================================================
# --------------------------------------------------------
# Forward transform
# --------------------------------------------------------
X = dft(g)

# --------------------------------------------------------
# Top-k threshold
# --------------------------------------------------------
k=40
Xk = top_k_threshold(
        X,
        k
    )

# Remove negligible imaginary part
Xt = np.real_if_close(Xk)

x_norm = Xt/np.linalg.norm(Xt)
# --------------------------------------------------------
# Reconstruction
# --------------------------------------------------------
rec = idft(x_norm)


# --------------------------------------------------------
# Metrics
# --------------------------------------------------------

TD = trace_distance(target_f, rec)
F = state_fidelity(target_f, rec)

print(
      f"k={k:3d} | "
      f"TD={TD:.6e} | "
      f"Fidelity={F:.10f}"
    )

# --------------------------------------------------------
# Plot
# --------------------------------------------------------

plt.figure(figsize=(8,5))
plt.plot(
        target_f,
        color="blue",
        lw=7,
        alpha=0.30,
        label="Original"
    )

plt.plot(
      rec,
        color="red",
        lw=2,
        label="SD-QSP (Fourier)"
    )
    

plt.title(f"Gaussian Mixture | k={k} |  TD={TD:.4e}")


plt.xlabel("Sample index")
plt.ylabel("Amplitude")

plt.grid(True)
plt.legend()
plt.savefig("../figures/MG_dft.pdf")

plt.tight_layout()
plt.show()
```

```python
vec = x_norm

nz = np.flatnonzero(vec)             # Nonzero indices

y = [(format(i, f"0{n}b"), vec[i]) for i in nz]

# ------------------------------------------------
# Sparse encoder circuit
# ------------------------------------------------
circuit = sparse_encoder(y, method="farias")

# ------------------------------------------------
# inverse QFT
# ------------------------------------------------
iqft = QFT(nqubits=n).invert()

circuit.add(iqft.on_qubits(*range(n)))

print(circuit.decompose().decompose().summary())
```
